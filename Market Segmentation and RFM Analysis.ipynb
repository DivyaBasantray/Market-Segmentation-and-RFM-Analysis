# RFM

```python
# Combine Recency, Frequency and Monetary into one table

rfm = pd.DataFrame({
    'Recency': recency,
    'Frequency': frequency,
    'Monetary': monetary
})

rfm.head()
```

```python
rfm.shape
```

```python
rfm.describe()
```

```python
rfm['R_Score'] = pd.qcut(
    rfm['Recency'],
    5,
    labels=[5, 4, 3, 2, 1]
)

rfm[['Recency', 'R_Score']].head(10)
```

```python
rfm['F_Score'] = pd.qcut(
    rfm['Frequency'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm[['Frequency', 'F_Score']].head(10)
```

```python
rfm['M_Score'] = pd.qcut(
    rfm['Monetary'],
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm[['Monetary', 'M_Score']].head(10)
```

```python
# Create the RFM code, e.g. 255 = R:2, F:5, M:5

rfm['RFM_Code'] = (
    rfm['R_Score'].astype(str) +
    rfm['F_Score'].astype(str) +
    rfm['M_Score'].astype(str)
)

# Calculate the total RFM score

rfm['RFM_Score'] = (
    rfm['R_Score'].astype(int) +
    rfm['F_Score'].astype(int) +
    rfm['M_Score'].astype(int)
)

rfm.head(10)
```

```python
# Count customers with high scores across all three dimensions

champions = rfm[
    (rfm['R_Score'] >= 4) &
    (rfm['F_Score'] >= 4) &
    (rfm['M_Score'] >= 4)
]

champions.shape
```

```python
# Customers who are less recent but have strong historical frequency or spending

at_risk = rfm[
    (rfm['R_Score'] <= 2) &
    (
        (rfm['F_Score'] >= 4) |
        (rfm['M_Score'] >= 4)
    )
]

at_risk.shape
```

```python
# Recent customers with relatively low frequency and spending

potential = rfm[
    (rfm['R_Score'] >= 4) &
    (rfm['F_Score'] <= 2) &
    (rfm['M_Score'] <= 2)
]

potential.shape
```

```python
# Low engagement and low spending

inactive = rfm[
    (rfm['R_Score'] <= 2) &
    (rfm['F_Score'] <= 2) &
    (rfm['M_Score'] <= 2)
]

inactive.shape
```

```python
# Create flags for each candidate segment

rfm['Champions_Flag'] = (
    (rfm['R_Score'] >= 4) &
    (rfm['F_Score'] >= 4) &
    (rfm['M_Score'] >= 4)
)

rfm['At_Risk_Flag'] = (
    (rfm['R_Score'] <= 2) &
    (
        (rfm['F_Score'] >= 4) |
        (rfm['M_Score'] >= 4)
    )
)

rfm['Potential_Flag'] = (
    (rfm['R_Score'] >= 4) &
    (rfm['F_Score'] <= 2) &
    (rfm['M_Score'] <= 2)
)

rfm['Inactive_Flag'] = (
    (rfm['R_Score'] <= 2) &
    (rfm['F_Score'] <= 2) &
    (rfm['M_Score'] <= 2)
)

# Count how many segment conditions each customer satisfies

rfm['Segment_Count'] = (
    rfm['Champions_Flag'].astype(int) +
    rfm['At_Risk_Flag'].astype(int) +
    rfm['Potential_Flag'].astype(int) +
    rfm['Inactive_Flag'].astype(int)
)

rfm['Segment_Count'].value_counts().sort_index()
```

```python
def assign_segment(row):

    R = row['R_Score']
    F = row['F_Score']
    M = row['M_Score']

    # 1. Champions
    if R >= 4 and F >= 4 and M >= 4:
        return 'Champions'

    # 2. At Risk
    elif R <= 2 and (F >= 4 or M >= 4):
        return 'At Risk'

    # 3. Potential Customers
    elif R >= 4 and F <= 2 and M <= 2:
        return 'Potential Customers'

    # 4. Inactive / Low Value
    elif R <= 2 and F <= 2 and M <= 2:
        return 'Inactive / Low Value'

    # 5. Remaining customers
    else:
        return 'Other Customers'


rfm['Segment'] = rfm.apply(assign_segment, axis=1)
```

```python
def assign_segment(row):

    R = row['R_Score']
    F = row['F_Score']
    M = row['M_Score']

    # 1. Champions
    if R >= 4 and F >= 4 and M >= 4:
        return 'Champions'

    # 2. At Risk
    elif R <= 2 and (F >= 3 or M >= 3):
        return 'At Risk'

    # 3. Loyal Customers
    elif R >= 3 and F >= 4:
        return 'Loyal Customers'

    # 4. Potential Customers
    elif R >= 4 and F <= 2:
        return 'Potential Customers'

    # 5. Inactive / Low Value
    elif R <= 2 and F <= 2 and M <= 2:
        return 'Inactive / Low Value'

    # 6. Regular / Mid-Value
    else:
        return 'Regular / Mid-Value'


rfm['Segment'] = rfm.apply(assign_segment, axis=1)
```

```python
segment_product_analysis = (
    df_products
    .groupby(['Segment', 'Description'])
    .agg(
        Revenue=('Revenue', 'sum'),
        Quantity=('Quantity', 'sum'),
        Customers=('Customer ID', 'nunique')
    )
    .reset_index()
)

segment_product_analysis.head()
```

```python
top_products_by_segment = (
    segment_product_analysis
    .sort_values(
        ['Segment', 'Revenue'],
        ascending=[True, False]
    )
    .groupby('Segment')
    .head(10)
)

top_products_by_segment
```

```python
segment_product_analysis['Revenue_per_Customer'] = (
    segment_product_analysis['Revenue'] /
    segment_product_analysis['Customers']
)
```

```python
top_revenue_per_customer = (
    segment_product_analysis
    .sort_values(
        ['Segment', 'Revenue_per_Customer'],
        ascending=[True, False]
    )
    .groupby('Segment')
    .head(10)
)

top_revenue_per_customer
```

```python
top_products_by_customer_reach = (
    segment_product_analysis
    .sort_values(
        ['Segment', 'Customers', 'Revenue'],
        ascending=[True, False, False]
    )
    .groupby('Segment')
    .head(10)
)

top_products_by_customer_reach
```

```python
segment_summary = (
    df_segmented.groupby('Segment')
    .agg(
        Customers=('Customer ID', 'nunique'),
        Revenue=('Revenue', 'sum'),
        Orders=('Invoice', 'nunique'),
        Recency=('Customer ID', lambda x: rfm.loc[x.unique(), 'Recency'].mean()),
        Frequency=('Customer ID', lambda x: rfm.loc[x.unique(), 'Frequency'].mean()),
        Monetary=('Customer ID', lambda x: rfm.loc[x.unique(), 'Monetary'].mean())
    )
    .reset_index()
)

segment_summary['Revenue_per_Customer'] = (
    segment_summary['Revenue'] / segment_summary['Customers']
)

segment_summary['Orders_per_Customer'] = (
    segment_summary['Orders'] / segment_summary['Customers']
)

segment_summary['Average_Order_Value'] = (
    segment_summary['Revenue'] / segment_summary['Orders']
)

segment_summary.sort_values('Revenue', ascending=False)
```
